```python
a = list(map(int, input().split()))
counts = {}

for x in a:
    if x not in counts:
        counts[x] = 0
    counts[x] += 1

xoa = 0
for x in counts:
    if x == 0:
        xoa += counts[x]
    elif counts[x] < x:
        xoa += counts[x]
    else:
        xoa += counts[x] - x

if xoa == 0:
    print("Mang dep")
else:
    print(xoa)
```

```python
a = list(map(int, input().split()))

candidate = None
count = 0

for x in a:
    if count == 0:
        candidate = x
        count = 1
    elif x == candidate:
        count += 1
    else:
        count -= 1

if candidate is not None:
    dem = 0
    for x in a:
        if x == candidate:
            dem += 1
    if dem > len(a) / 2:
        print(candidate)
```

```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))

if len(a) >= 2 and a[0] > a[-1]:
    i = len(a) - 1
    step_a = -1
else:
    i = 0
    step_a = 1

if len(b) >= 2 and b[0] > b[-1]:
    j = len(b) - 1
    step_b = -1
else:
    j = 0
    step_b = 1

ket_qua = []

while 0 <= i < len(a) and 0 <= j < len(b):
    if a[i] <= b[j]:
        ket_qua.append(a[i])
        i += step_a
    else:
        ket_qua.append(b[j])
        j += step_b

while 0 <= i < len(a):
    ket_qua.append(a[i])
    i += step_a

while 0 <= j < len(b):
    ket_qua.append(b[j])
    j += step_b

print(*ket_qua)
```

```python
n = int(input())
counts = {}

for i in range(n):
    full_name = input().strip()
    parts = full_name.split()
    if len(parts) > 0:
        ten = parts[-1]
        if ten not in counts:
            counts[ten] = 0
        counts[ten] += 1

best_name = ""
best_count = -1

for ten in counts:
    if counts[ten] > best_count or (counts[ten] == best_count and ten < best_name):
        best_name = ten
        best_count = counts[ten]

print(best_name)
```

```python
s = input().strip()
k = int(input())

can_xoa = len(s) - k
stack = []

for c in s:
    while can_xoa > 0 and len(stack) > 0 and stack[-1] < c:
        stack.pop()
        can_xoa -= 1
    stack.append(c)

while can_xoa > 0 and len(stack) > 0:
    stack.pop()
    can_xoa -= 1

print("".join(stack[:k]))
```

```python
s = input().strip()
stack = []
pairs = {')': '(', ']': '[', '}': '{'}
hop_le = True

for c in s:
    if c in "([{":
        stack.append(c)
    elif c in ")]}":
        if len(stack) == 0 or stack[-1] != pairs[c]:
            hop_le = False
            break
        stack.pop()

if len(stack) != 0:
    hop_le = False

if hop_le:
    print("Hop le")
else:
    print("Khong hop le")
```

```python
def is_composite(x):
    if x <= 1:
        return False
    if x == 2:
        return False
    i = 2
    while i * i <= x:
        if x % i == 0:
            return True
        i += 1
    return False


N = int(input())
best_start = -1
best_len = 0
cur_start = -1
cur_len = 0

for x in range(2, N + 1):
    if is_composite(x):
        if cur_len == 0:
            cur_start = x
        cur_len += 1
    else:
        if cur_len > best_len:
            best_start = cur_start
            best_len = cur_len
        cur_len = 0

if cur_len > best_len:
    best_start = cur_start
    best_len = cur_len

if best_len == 0:
    print(-1)
else:
    for x in range(best_start, best_start + best_len):
        print(x, end=" ")
```

```python
a = list(map(int, input().split()))

best_start = 0
best_len = 0
cur_start = 0
cur_len = 0

if len(a) > 0:
    best_len = 1
    cur_len = 1

for i in range(1, len(a)):
    if a[i] > a[i - 1]:
        cur_len += 1
    else:
        if cur_len > best_len:
            best_start = cur_start
            best_len = cur_len
        cur_start = i
        cur_len = 1

if cur_len > best_len:
    best_start = cur_start
    best_len = cur_len

print(*a[best_start:best_start + best_len])
```

```python
a = list(map(int, input().split()))
n = len(a)

if n == 0:
    print()
else:
    dp = [1] * n
    prev = [-1] * n

    for i in range(n):
        for j in range(i):
            if a[j] < a[i] and dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1
                prev[i] = j

    best = 0
    for i in range(1, n):
        if dp[i] > dp[best]:
            best = i

    result = []
    while best != -1:
        result.append(a[best])
        best = prev[best]

    result.reverse()
    print(*result)
```

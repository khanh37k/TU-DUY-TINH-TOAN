```python
n = int(input())

chan = 0
le = 0

if n == 0:
    chan = 1
else:
    while n > 0:
        d = n % 10
        if d % 2 == 0:
            chan += 1
        else:
            le += 1
        n //= 10

print("So chu so chan:", chan)
print("So chu so le:", le)
```

```python
x = int(input())

if x == -1:
    print("Khong co so nao")
else:
    lon_nhat = x
    nho_nhat = x

    while True:
        x = int(input())
        if x == -1:
            break
        if x > lon_nhat:
            lon_nhat = x
        if x < nho_nhat:
            nho_nhat = x

    print("So lon nhat:", lon_nhat)
    print("So nho nhat:", nho_nhat)
```

```python
A = int(input())

if A < 1:
    print(0)
else:
    a = 1
    b = 1
    lon_nhat = 1

    while b <= A:
        lon_nhat = b
        c = a + b
        a = b
        b = c

    print(lon_nhat)
```

```python
s = input()

n = 0
for c in s:
    n += 1

dem_tu = 0
trong_tu = False
for i in range(n):
    if s[i] != " " and not trong_tu:
        dem_tu += 1
        trong_tu = True
    elif s[i] == " ":
        trong_tu = False

print(dem_tu)
```

```python
s = input()

n = 0
for c in s:
    n += 1

tu_dau = ""
da_bat_dau = False
for i in range(n):
    if s[i] != " ":
        tu_dau += s[i]
        da_bat_dau = True
    elif da_bat_dau:
        break

print(tu_dau)
```

```python
s = input()

ket_qua = ""
for c in s:
    if "a" <= c <= "z":
        ket_qua += chr(ord(c) - 32)
    elif "A" <= c <= "Z":
        ket_qua += chr(ord(c) + 32)
    else:
        ket_qua += c

print(ket_qua)
```

```python
s = input()

tong = 0
so = 0
dau = 1
dang_doc_so = False

for c in s + ",":
    if c == " ":
        continue
    elif c == "-":
        dau = -1
    elif "0" <= c <= "9":
        so = so * 10 + ord(c) - ord("0")
        dang_doc_so = True
    elif c == ",":
        if dang_doc_so:
            tong += dau * so
        so = 0
        dau = 1
        dang_doc_so = False

print(tong)
```

```python
s = input()

hoa = 0
thuong = 0
so = 0

for c in s:
    if "A" <= c <= "Z":
        hoa += 1
    elif "a" <= c <= "z":
        thuong += 1
    elif "0" <= c <= "9":
        so += 1

print("Ky tu in hoa:", hoa)
print("Ky tu in thuong:", thuong)
print("Ky tu so:", so)
```

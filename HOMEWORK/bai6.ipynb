```python
def remove_less_than(a, x):
    i = 0
    while i < len(a):
        if a[i] < x:
            a.pop(i)
        else:
            i += 1

    return a
```

```python
def move_zero_to_front(a):
    zeros = []
    others = []

    for x in a:
        if x == 0:
            zeros.append(x)
        else:
            others.append(x)

    return zeros + others


def move_zero_to_back(a):
    zeros = []
    others = []

    for x in a:
        if x == 0:
            zeros.append(x)
        else:
            others.append(x)

    return others + zeros
```

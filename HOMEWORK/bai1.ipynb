```python
#Bai 6:
import math

a, b, c = map(float, input().split())

if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    s = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print(round(s, 1))
else:
    print("Khong phai 3 canh cua tam giac")
```

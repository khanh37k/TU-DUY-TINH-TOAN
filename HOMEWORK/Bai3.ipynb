```python
h = int(input())

if h <= 0:
    print("Chieu cao khong hop le")
elif h == 1:
    print("*")
else:
    i = 1
    while i <= h:
        space = 1
        while space <= h - i:
            print(" ", end="")
            space += 1

        if i == 1:
            print("*")
        elif i == h:
            star = 1
            while star <= 2 * h - 1:
                print("*", end="")
                star += 1
            print()
        else:
            print("*", end="")
            mid = 1
            while mid <= 2 * i - 3:
                print(" ", end="")
                mid += 1
            print("*")

        i += 1
```

```python
x = int(input())

if x == -1:
    print("Khong co so nao")
else:
    lon_nhat = x
    nho_nhat = x

    while True:
        x = int(input())
        if x == -1:
            break
        if x > lon_nhat:
            lon_nhat = x
        if x < nho_nhat:
            nho_nhat = x

    print("So lon nhat:", lon_nhat)
    print("So nho nhat:", nho_nhat)
```

```python
n = int(input())

chan = 0
le = 0

if n == 0:
    chan = 1
else:
    while n > 0:
        d = n % 10
        if d % 2 == 0:
            chan += 1
        else:
            le += 1
        n //= 10

print("So chu so chan:", chan)
print("So chu so le:", le)
```

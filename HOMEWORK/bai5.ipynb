```python
def is_armstrong(x):
    if x < 0:
        return False

    n = x
    dem = 1 if n == 0 else 0
    temp = n
    while temp > 0:
        dem += 1
        temp //= 10

    tong = 0
    temp = n
    if temp == 0:
        tong = 0
    while temp > 0:
        d = temp % 10
        tong += d ** dem
        temp //= 10

    return tong == n
```

```python
def is_square(x):
    if x < 0:
        return False

    i = 0
    while i * i < x:
        i += 1

    return i * i == x
```

```python
def is_palindrome_number(x):
    if x < 0:
        return False

    n = x
    dao = 0
    while n > 0:
        dao = dao * 10 + n % 10
        n //= 10

    return dao == x
```

```python
def is_magic_number(x):
    if x <= 0:
        return False

    while x >= 10:
        tong = 0
        while x > 0:
            tong += x % 10
            x //= 10
        x = tong

    return x == 1
```

```python
def detect_number_type(s):
    s = s.strip()
    if s == "":
        return "neither"

    try:
        int(s)
        return "int"
    except ValueError:
        pass

    try:
        value = float(s)
        if value != value or value == float("inf") or value == float("-inf"):
            return "neither"
        return "float"
    except ValueError:
        return "neither"
```

```python
def is_one_product_of_others(a, b, c):
    return a == b * c or b == a * c or c == a * b
```

```python
def sum_halves(s):
    return int(s[:3]) + int(s[3:])
```

```python
def quadratic_roots_count(a, b, c):
    if a == 0:
        if b == 0:
            if c == 0:
                return -1
            return 0
        return 1

    delta = b * b - 4 * a * c
    if delta > 0:
        return 2
    if delta == 0:
        return 1
    return 0
```

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (x2 - x1) * (y3 - y1) == (y2 - y1) * (x3 - x1)
```

```python
def _point_on_segment(a, b, p):
    x1, y1 = a
    x2, y2 = b
    x, y = p
    cross = (x - x1) * (y2 - y1) - (y - y1) * (x2 - x1)
    if abs(cross) > 1e-9:
        return False
    return min(x1, x2) <= x <= max(x1, x2) and min(y1, y2) <= y <= max(y1, y2)


def point_in_quadrilateral(quad, p):
    x, y = p
    inside = False
    j = 3

    for i in range(4):
        a = quad[i]
        b = quad[j]
        if _point_on_segment(a, b, p):
            return True

        xi, yi = a
        xj, yj = b
        if (yi > y) != (yj > y):
            x_intersect = (xj - xi) * (y - yi) / (yj - yi) + xi
            if x < x_intersect:
                inside = not inside
        j = i

    return inside
```

```python
def count_nonzero_digits(numbers):
    dem = 0
    for n in numbers:
        n = abs(n)
        while n > 0:
            if n % 10 != 0:
                dem += 1
            n //= 10
    return dem
```

```python
def angle_abc(x1, y1, x2, y2, x3, y3):
    import math

    bax = x1 - x2
    bay = y1 - y2
    bcx = x3 - x2
    bcy = y3 - y2

    dot = bax * bcx + bay * bcy
    len_ba = math.sqrt(bax * bax + bay * bay)
    len_bc = math.sqrt(bcx * bcx + bcy * bcy)

    if len_ba == 0 or len_bc == 0:
        return None

    cos_angle = dot / (len_ba * len_bc)
    if cos_angle > 1:
        cos_angle = 1
    elif cos_angle < -1:
        cos_angle = -1

    return math.degrees(math.acos(cos_angle))
```

```python
def max_min_three(a, b, c):
    lon_nhat = a
    nho_nhat = a

    if b > lon_nhat:
        lon_nhat = b
    if c > lon_nhat:
        lon_nhat = c

    if b < nho_nhat:
        nho_nhat = b
    if c < nho_nhat:
        nho_nhat = c

    return [lon_nhat, nho_nhat]
```

```python
def binomial_coefficient(n, k):
    if k < 0 or k > n:
        return 0

    if k > n - k:
        k = n - k

    result = 1
    for i in range(1, k + 1):
        result = result * (n - k + i) // i

    return result
```

```python
def sum_of_squares(n):
    tong = 0
    for i in range(1, n + 1):
        tong += i * i
    return tong
```

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    rates = {
        "4": {"open": 12000, "km": 15000, "wait": 1000},
        "7": {"open": 15000, "km": 18000, "wait": 1200},
    }

    car_type = str(car_type)
    if car_type not in rates or km < 0 or wait_minutes < 0 or start_hour < 0 or start_hour > 23:
        return None

    rate = rates[car_type]
    fare = rate["open"]
    if km > 1:
        fare += (km - 1) * rate["km"]

    fare += wait_minutes * rate["wait"]

    if start_hour >= 22 or start_hour < 5:
        fare *= 1.2

    if km >= 30:
        fare *= 0.9

    return round(fare)
```

```python
def _different_digits(x):
    x = abs(x)
    used = set()

    if x == 0:
        return True

    while x > 0:
        d = x % 10
        if d in used:
            return False
        used.add(d)
        x //= 10

    return True


def sum_1_to_n(n):
    if n <= 0:
        return 0

    tong = 0
    for x in range(1, n + 1):
        if _different_digits(x):
            tong += x

    return tong
```
